```python
import itertools
import re
import string
```

```python
EN_US = 0
ZH_HANS = 1

def langs_of(s):
    if re.fullmatch(r'[a-zA-Z0-9]?', s):
        return EN_US, ZH_HANS
    elif re.fullmatch(r'[A-Za-z0-9]+', s):
        return EN_US,
    else:
        return ZH_HANS,
```

```python
L = ['L', 'Left', '左']
R = ['R', 'Right', '右']
M = ['M', 'Middle', '中']
```

```python
#TODO: Multimedia keys
#TODO: Virtual key code and scan code
keyboard_keys = {
    # Best names: alternative names
    # (CODE, EN_US, ZH_HANS): [[part1...], [part2...], ...]
}

for s in string.ascii_uppercase:
    keyboard_keys[(s,)] = [[s]]

for s in string.digits:
    keyboard_keys[(s,)] = [[s]]

F = ['F']
for i in range(1, 25):
    keyboard_keys[('F' + str(i),)] = [F, [str(i)]]

Numpad = ['Num', 'NumPad', 'NumberPad', '小键盘']
Hyphen = ['-', 'Substract', '减']
Addtion = ['+', 'Add', 'Addtion', '加']
Dot = ['.', 'Dot', '小数点', '句号', '句点']
Ctrl = ['Ctrl', 'Control', '控制']
Shift = ['Shift', '转换']
Alt = ['Alt']
Win = ['Win']
Enter = ['Enter', '回车']
Insert = ['Ins', 'Insert', '插入']
Delete = ['Del', 'Delete', '删除']
Home = ['Home']
End = ['End']
Page = ['Page']
PageDown = ['PageDown']
PageUp = ['PageUp']

Up = ['↑', 'Up', '上', '向上']
Down = ['↓', 'Down', '下', '向下']
Left = ['←', 'Left', '左', '向左']
Right = ['→', 'Right', '右', '向右']

keyboard_keys |= {
    ('Shift1','!'): [['!']],
    ('Shift2','@'): [['@']],
    ('Shift3','#'): [['#']],
    ('Shift4','$'): [['$']],
    ('Shift5','%'): [['%']],
    ('Shift6','^'): [['^']],
    ('Shift7','&'): [['&']],
    ('Shift8','*'): [['*']],
    ('Shift9','('): [['(']],
    ('Shift0',')'): [[')']],

    ('Numpad0','Numpad0','小键盘0'): [Numpad, ['0']+Insert],
    ('Numpad1','Numpad1','小键盘1'): [Numpad, ['1']+End],
    ('Numpad2','Numpad2','小键盘2'): [Numpad, ['2']+Down],
    ('Numpad3','Numpad3','小键盘3'): [Numpad, ['3']+PageDown],
    ('Numpad4','Numpad4','小键盘4'): [Numpad, ['4']+Left],
    ('Numpad5','Numpad5','小键盘5'): [Numpad, ['5', 'Clear']],
    ('Numpad6','Numpad6','小键盘6'): [Numpad, ['6']+Right],
    ('Numpad7','Numpad7','小键盘7'): [Numpad, ['7']+Home],
    ('Numpad8','Numpad8','小键盘8'): [Numpad, ['8']+Up],
    ('Numpad9','Numpad9','小键盘9'): [Numpad, ['9']+PageUp],
    ('NumpadDot','Numpad.','小键盘.'): [Numpad, Dot+Delete],
    ('NumpadAdd','Numpad+','小键盘+'): [Numpad, Addtion],
    ('NumpadSub','Numpad-','小键盘-'): [Numpad, Hyphen],
    ('NumpadMul','Numpad*','小键盘*'): [Numpad, ['*', 'Mul', 'Mult', 'Multiply']],
    ('NumpadDiv','Numpad/','小键盘/'): [Numpad, ['/', 'Div', 'Divide', 'Devision']],
    ('NumpadEnter','NumpadEnter','小键盘回车'): [Numpad, Enter],
    ('NumLock','NumLock','小键盘锁定'): [Numpad, ['NumLock', '小键盘锁定']],

    ('Dot','.'): [Dot],

    ('Ctrl','Ctrl','Ctrl'): [Ctrl],
    ('Shift','Shift','Shift'): [Shift],
    ('Alt','Alt','Alt'): [Alt],
    ('Win','Win','Win'): [Win],
    ('LCtrl','LCtrl','左Ctrl'): [L, Ctrl],
    ('RCtrl','RCtrl','右Ctrl'): [R, Ctrl],
    ('LShift','LShift','左Shift'): [L, Shift],
    ('RShift','RShift','右Shift'): [R, Shift],
    ('LAlt','LAlt','左Alt'): [L, Alt],
    ('RAlt','RAlt','右Alt'): [R, Alt],
    ('LWin','LWin','左Win'): [L, Win],
    ('RWin','RWin','右Win'): [R, Win],

    ('CapsLock','CapsLock','大写锁定'): [['CapsLock', '大写', '大写锁定']],
    ('Space','Space','空格'): [['Space', '空格']],
    ('Tab','Tab','Tab'): [['Tab']],
    ('Enter','Enter','回车'): [Enter],
    ('Esc','Esc','Esc'): [['Esc', 'Escape', '返回']],
    ('Backspace','Backspace','退格'): [['Backspace', 'BackSp', 'BS', '退格']],

    ('ScrollLock','ScrollLock','滚动锁定'): [['ScrollLock', '滚动锁定']],

    ('Insert','Insert','插入'): [Insert],
    ('Delete','Delete','删除'): [Delete],

    ('Home','Home','Home'): [Home],
    ('End','End','End'): [End],

    ('Up','↑','↑'): [Up],
    ('Down','↓','↓'): [Down],
    ('Left','←','←'): [Left],
    ('Right','→','→'): [Right],

    ('PageUp','PageUp','PageUp'): [PageUp],
    ('PageDown','PageDown','PageDown'): [PageDown],

    ('AppsKey','AppsKey','AppsKey'): [['AppsKey', 'Menu', '菜单']],
    ('PrintScreen','PrintScreen','截屏'): [['PrintScreen', 'PrintScr', '截屏']],
    ('Pause','Pause','暂停'): [['Pause', '暂停']],
    ('Sleep','Sleep','Sleep'): [['Sleep', '睡眠']],
    ('Help','Help','帮助'): [['Help', '帮助']],

    ('Grave','`'): [['`', 'Grave']],
    ('ShiftGrave','~'): [['~']],

    ('Hyphen','-'): [Hyphen],
    ('ShiftHyphen',): [['_']],

    ('Equal','='): [['=', 'Euqal']],
    ('ShiftEqual','+'): [Addtion],
}

Keyboard = ['', 'Keyboard', '键盘']
Key = ['', 'Key', '键']
for k in keyboard_keys:
    keyboard_keys[k].insert(0, Keyboard)
    keyboard_keys[k].append(Key)

Mouse = ['', 'Mouse', '鼠标']
Button = ['Btn', 'Button', '键']
X = ['X', '侧']
Wheel = ['Wheel', '滚轮']
mouse_keys = {
    ('LButton','LButton','左键'): [L, Button],
    ('RButton','RButton','右键'): [R, Button],
    ('MButton','MButton','中键'): [M, Button],

    ('X1Button','X1Button','侧键1'): [X, '1', Button],
    ('X2Button','X2Button','侧键2'): [X, '2', Button],

    ('WheelUp','WheelUp','滚轮上'): [Wheel, Up],
    ('WheelDown','WheelDown','滚轮下'): [Wheel, Down],
    ('WheelLeft','WheelLeft','滚轮左'): [Wheel, Left],
    ('WheelRight','WheelRight','滚轮右'): [Wheel, Right],
}
for k in mouse_keys:
    mouse_keys[k].insert(0, Mouse)

keys = keyboard_keys | mouse_keys
```

```python
reverse_dict = {}
for k, lst in keys.items():
    print(f'{k[0]}: {list(k)}')
    for lang in EN_US, ZH_HANS:
        product = itertools.product(*[
            [s for s in filter(lambda x: lang in langs_of(x), part)]
            for part in lst])
        s = [''.join(s) for s in product]
        print(f"{k[0]}: {s}")

        for s in s:
            reverse_dict[s.lower()] = k[0]

#print(reverse_dict)
```
